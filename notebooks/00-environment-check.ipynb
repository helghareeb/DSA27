## 3. Stepping through an algorithm

Write the algorithm as a **generator** that yields a snapshot after each step. That one `yield` is what makes every intermediate state inspectable.

```python
def bubble_sort_steps(values):
    values = list(values)
    for i in range(len(values)):
        for j in range(len(values) - i - 1):
            yield values, (j, j + 1)          # about to compare these two
            if values[j] > values[j + 1]:
                values[j], values[j + 1] = values[j + 1], values[j]
    yield values, ()


frames = list(bubble_sort_steps([5, 2, 9, 1, 7]))
print(f"{len(frames)} steps; final state {frames[-1][0]}")
```

## 4. Measuring complexity

Don't assert that bubble sort is $O(n^2)$ — measure it, and lay $n$ and $n^2$ over the result.

```python
import random
from viz.complexity import measure, plot_growth

def bubble_sort(values):
    for state, _ in bubble_sort_steps(values):
        pass
    return state

sizes = [100, 200, 400, 800]
make = lambda n: [random.randint(0, 1000) for _ in range(n)]

plot_growth({"bubble sort": measure(bubble_sort, sizes, make, repeat=2)},
            reference=["n", "n^2"], title="Bubble sort really is quadratic");
```
