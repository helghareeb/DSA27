```python
from viz.complexity import measure, plot_growth
import random

def linear_search(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index
    return -1

sizes = [1000, 2000, 4000, 8000, 16000]
make = lambda n: (sorted(random.random() for _ in range(n)), 2.0)

plot_growth({"linear": measure(lambda a: linear_search(*a), sizes, make)},
            reference=["n"])
```
